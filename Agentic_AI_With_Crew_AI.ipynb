# Create Project

- A prompt is going to apear and will ask you to select a provider (e.g. openai, gemini, ...)
- You need to select Number (Gemini)
- 2nd prompt will ask you to select model. You can pick any number here. We are going to change it later to `gemini/gemini-2.5-flash`
- The 3rd step is to provide a GEMINI API key. You can see how [here](https://www.youtube.com/watch?v=yZN5a12CZD8)


```python
# To prevent your API key from appearing in the notebook output when creating a CrewAI project:
#
# 1. When prompted "Enter your [PROVIDER] API key... (press Enter to skip):", simply press Enter.
#    This will create the project structure without embedding the key in the output.
#
# 2. After the project is created, open the generated `.env` file located in your project directory
#    (e.g., `/content/hello_world/.env`).
#
# 3. Add or update the line for your API key in the `.env` file. For Gemini, it would look like this:
#    GEMINI_API_KEY='YOUR_ACTUAL_GEMINI_API_KEY_HERE'
#    Replace 'YOUR_ACTUAL_GEMINI_API_KEY_HERE' with your actual key.

!uv tool run crewai create crew hello_world
```
